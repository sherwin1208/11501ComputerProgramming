### Part 1：完成函數

請閱讀以下函數，理解各部分的功能。

```python
def analyze_scores(*scores, **options):
    """
    分析學生分數。

    Parameters:
        *scores: 接收不定數量的分數
        **options: 接收選擇性設定，例如 multiplier 和 bonus

    Returns:
        一個字典，包含調整後的分數、平方、總分及平均分數。
    """
    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    adjusted = [
        score * multiplier + bonus
        for score in scores
    ]

    squared = [
        score ** 2
        for score in adjusted
    ]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average
    }
```

## 三、語法整理（Learning Focus）

| 語法 | 用途 |
|---|---|
| `*scores`（函數定義） | 將不定數量的位置參數收集成 tuple |
| `**options`（函數定義） | 將不定數量的關鍵字參數收集成 dictionary |
| `score * multiplier` | 乘法 |
| `score ** 2` | 次方 |
| `f(*scores)` | 將可迭代物件解包為位置參數 |
| `f(**settings)` | 將字典解包為關鍵字參數 |
| `return` | 將函數的結果回傳給呼叫端 |

在 Python 中，*、** 與 return 是非常核心且功能多樣的語法。以下根據本題的脈絡與 Python 的特性，為您詳細說明它們在不同情境下的用途，並透過程式碼範例展示「同一個符號在不同語法位置所代表的意義」。

一、 *（星號）的三種主要用途
數學運算（乘法）：

語法位置：用在兩個數值之間（例如：score * multiplier）。

意義：進行數值相乘。

定義函數時的打包（Packing）：

語法位置：用在函數定義的參數前（例如：def analyze_scores(*scores)）。

意義：接收不定數量的位置參數，並將它們打包成一個 Tuple（元組）。

呼叫函數或容器時的解包（Unpacking）：

語法位置：用在呼叫函數或建立串列時的容器前（例如：analyze_scores(*scores)）。

意義：將串列或元組中的元素「拆開」，逐一傳遞給函數作為獨立的位置參數。

二、 **（雙星號）的三種主要用途
數學運算（次方 / 冪）：

語法位置：用在兩個數值之間（例如：score ** 2）。

意義：計算次方（例如 2 ** 3 代表 2 的 3 次方，結果為 8）。

定義函數時的打包（Packing）：

語法位置：用在函數定義的參數前（例如：def analyze_scores(**options)）。

意義：接收不定數量的關鍵字參數，並將它們打包成一個 Dict（字典）。

呼叫函數時的解包（Unpacking）：

語法位置：用在呼叫函數時的字典前（例如：analyze_scores(**settings)）。

意義：將字典的鍵值對（Key-Value）「拆開」，作為獨立的關鍵字引數傳入函數中。

三、 return 的用途
語法位置：僅能使用在函數（Function）內部。

意義：

終止函數執行：當程式執行到 return 時，該函數會立刻結束，後續的程式碼不會被執行。

回傳資料：將運算結果（可以是數值、字串、串列、字典等）送回給函數的呼叫端，讓外部的變數可以接收並繼續利用這些資料。

四、 程式碼範例：同一個符號在不同語法位置的意義
以下用程式碼實際展示 * 與 ** 在不同位置時的雙重身分：
# ==========================================
# 範例 1：* 號的雙重身分（乘法 vs. 解包）
# ==========================================

def multiply_two(a, b):
    # 這裡的 * 是「數學運算：乘法」
    return a * b

# 建立一個串列
numbers = [6, 7]

# 這裡的 * 是「解包（Unpacking）」
# 它把 numbers 裡的 6 和 7 拆開，變成傳入 multiply_two(6, 7)
result = multiply_two(*numbers)
print("乘法運算與解包結果：", result)  # 輸出：42


# ==========================================
# 範例 2：** 號的三重身分（次方 vs. 打包 vs. 解包）
# ==========================================

# 1. 數學運算：次方
base = 3
exponent = 4
power_val = base ** exponent  # 這裡的 ** 代表 3 的 4 次方 (3^4 = 81)
print("次方運算結果：", power_val)


# 2. 定義函數時的「打包（Packing）」
def show_config(**kwargs):
    # 這裡的 ** 把傳入的關鍵字參數打包成字典
    print("打包後的字典：", kwargs)
    return kwargs


# 3. 呼叫函數時的「解包（Unpacking）」
my_settings = {"theme": "dark", "version": 2.0}

# 這裡的 ** 把字典解包成 theme="dark", version=2.0 傳入函數
show_config(**my_settings)
