習題一：使用random模組，產生出一個介於0到1之間的小數$r$，使用while迴圈，計算出$r$的前二十位二進位表示$r = 0.b_1b_2b_3\cdots {b_{20}}_2$。輸出格式如附圖Exercise20261015.png

import random

# 1. 產生 0 到 1 之間的小數 r
r = random.random()
# 為了儲存原本的 r 值供最後輸出顯示，先備份一份
original_r = r

# 2. 初始化二進位字串與計數器
binary_bits = ""
count = 0

# 3. 使用 while 迴圈計算前 20 位二進位
while count < 20:
    r *= 2             # 將小數乘以 2
    bit = int(r)       # 取出整數部分 (0 或 1)
    binary_bits += str(bit)
    r -= bit           # 減去整數部分，保留新的小數部分
    count += 1         # 次數加 1

# 4. 格式化輸出結果
print(f"假如 r = {original_r}")
print(f"r ≈ 0.{binary_bits}_2")

Python 的 `int` 型別採用 Arbitrary-Precision Integer（任意精度整數）設計，能夠精確表示任意大小的整數，其實作方式可參考 CPython 的 `PyLongObject`。與 C/C++ 的 `int` 或 `long long` 不同，Python 的整數不受固定 32 位元或64 位元限制，而是會依據數值大小自動配置更多記憶體，因此只要記憶體足夠，便能表示極大的整數。例如：
```python
n = 10**68
print(n)
```
上述程式可正常執行，表示 Python 能夠精確儲存$10^{68}$，即中文數字中的「無量大數」。


可利用下列程式觀察不同大小整數所需的位元數（bit）與記憶體空間：

```python
import sys
n = 10**68
print(f"bit_length = {n.bit_length()}")
print(f"memory = {sys.getsizeof(n)} bytes")
```

習題三：請勿使用for、range，而是利用iterable、iter()、next()、while、StopIteration來設計九九乘法表。

# 建立外層與內層數字的迭代器
outer_iter = iter([1, 2, 3, 4, 5, 6, 7, 8, 9])

while True:
    try:
        # 取出外層的數字 (i)
        i = next(outer_iter)
        
        # 每次外層重頭開始時，重新建立內層數字的迭代器
        inner_iter = iter([1, 2, 3, 4, 5, 6, 7, 8, 9])
        
        while True:
            try:
                # 取出內層的數字 (j)
                j = next(inner_iter)
                print(f"{i}x{j}={i*j:2d}", end="  ")
            except StopIteration:
                # 內層數字取完後跳出內層迴圈，並換行
                print()
                break

    except StopIteration:
        # 外層數字取完後結束整個流程
        break
