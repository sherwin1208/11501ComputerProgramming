n = int(input())

for i in range(1, n + 1):
    for j in range(1, n + 1):
        product = i * j
        # 使用 f-string 的 :03b (3位二進位補0) 與 :06b (6位二進位補0)
        end_char = "  " if j < n else "\n"
        print(f"{i:03b}*{j:03b}={product:06b}", end=end_char)

使用者輸入：請輸入正整數$n$，其中$1\le n\le 9$，輸出二進位$nn$乘法表。格式如如Tranclass的圖片Exercise20261008.png，以$n=5$為例。                                                                                                         
                                                                                                            
1. 只能使用while。                                                                                            
3. 字串格式化只能使用 f-string 對齊。     

n = int(input())

i = 1
while i <= n:
    j = 1
    while j <= n:
        product = i * j
        # 使用 f-string 二進位格式化：03b (3位補零)、06b (6位補零)
        end_char = "  " if j < n else "\n"
        print(f"{i:03b}*{j:03b}={product:06b}", end=end_char)
        j += 1
    i += 1
